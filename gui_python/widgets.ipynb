# Exemplos de aplicações GUI

```python
import tkinter as tk
from tkinter import messagebox as mb
from tkinter import ttk
```

# Widgets

```python
window = tk.Tk()
window.geometry("500x500")
window.title("Aplicação GUI")
window.mainloop()
```

## Label

```python
window = tk.Tk()
window.geometry("500x500")
window.title("Aplicação GUI com o Widget Label")
ttk.Label(window, text="Componente Label").grid(column=0, row=0)
window.mainloop()
```

## Button

```python
contador = 0


def contador_label(lbl_rotulo: tk.Label):
    def funcao_contar():
        global contador
        contador += 1
        lbl_rotulo.config(text=str(contador))
        lbl_rotulo.after(1000, funcao_contar)

    funcao_contar()


window = tk.Tk()
window.title("Contagem dos Segundos")

lblRotulo = tk.Label(window, fg="green")
lblRotulo.pack()

contador_label(lblRotulo)

btn_acao = tk.Button(window, text='Clique aqui para Interromper a contagem', width=50, command=window.destroy)
btn_acao.pack()

window.mainloop()
```

## Entry

```python
def mostrar_nomes():
    print("Nome: %s\nSobrenome: %s" % (e1.get(), e2.get()))


window = tk.Tk()
window.title("Aplicação GUI com o Widget Entry")
tk.Label(window, text="Nome").grid(row=0)

tk.Label(window, text="Sobrenome").grid(row=1)
e1 = tk.Entry(window)
e2 = tk.Entry(window)
e1.grid(row=0, column=1)
e2.grid(row=1, column=1)
tk.Button(window, text='Sair', command=window.quit).grid(row=3, column=0, sticky=tk.W, pady=4)
tk.Button(window, text='Exibir Dados', command=mostrar_nomes).grid(row=3, column=1, sticky=tk.W, pady=4)
tk.mainloop()
```

## Radiobutton

```python
window = tk.Tk()
v = tk.IntVar()
tk.Label(window, text="""Escolha uma linguagem de programação:""", justify=tk.LEFT, padx=20).pack()
tk.Radiobutton(window, text="Python", padx=25, variable=v, value=1).pack(anchor=tk.W)
tk.Radiobutton(window, text="C++", padx=25, variable=v, value=2).pack(anchor=tk.W)
window.mainloop()
```

## Checkbutton

```python
def escolha_carreira():
    print("Gerencial: %d,\nTécnica : %d" % (var1.get(), var2.get()))


window = tk.Tk()
ttk.Label(window, text="Escolha sua vocação:").grid(row=0, sticky=tk.W)
var1 = tk.IntVar()
ttk.Checkbutton(window, text="Gerencial", variable=var1).grid(row=1, sticky=tk.W)
var2 = tk.IntVar()
ttk.Checkbutton(window, text="Técnica", variable=var2).grid(row=2, sticky=tk.W)
ttk.Button(window, text='Sair', command=window.quit).grid(row=3, sticky=tk.W, pady=4)
ttk.Button(window, text='Mostrar', command=escolha_carreira).grid(row=4, sticky=tk.W, pady=4)
window.mainloop()
```

## Text

```python
window = tk.Tk()
T = tk.Text(window, height=2, width=30)
T.pack()
T.insert(tk.END, "Este é um texto\ncom duas linhas\n")
tk.mainloop()
```

## Message

```python
window = tk.Tk()
mensagem_para_usuario = "Está é uma mensagem.\n(Pode ser bastante útil para o usuário)"
msg = tk.Message(window, text=mensagem_para_usuario)
msg.config(bg='lightgreen', font=('times', 24, 'italic'))
msg.pack()
window.mainloop()
```

## Slider

```python
def mostrar_valores() -> None:
    print(w1.get(), w2.get())


window = tk.Tk()
w1 = ttk.Scale(window, from_=0, to=50)
w1.pack()
w2 = ttk.Scale(window, from_=0, to=100, orient=tk.HORIZONTAL)
w2.pack()
ttk.Button(window, text='Mostrar a Escala', command=mostrar_valores).pack()
window.mainloop()
```

## Dialog

```python
def resposta():
    mb.showerror("Resposta", "Desculpe, nenhuma resposta disponível!")


def verificacao():
    if mb.askyesno('Verificar', 'Realmente quer sair?'):
        mb.showwarning('Yes', 'Ainda não foi implementado')
    else:
        mb.showinfo('No', 'A opção de Sair foi cancelada')


tk.Button(text='Sair', command=verificacao).pack(fill=tk.X)
tk.Button(text='Resposta', command=resposta).pack(fill=tk.X)
tk.mainloop()
```

## Combobox

```python
# Criação de uma janela tkinter
janela = tk.Tk()
janela.title('Combobox')
janela.geometry('500x250')
# Componente Label
ttk.Label(janela, text="Combobox Widget", background='green', foreground="white", font=(
"Times New Roman", 15)).grid(row=0, column=1)
# Componente Label
ttk.Label(janela, text="Selecione um mês :", font=("Times New Roman", 10)).grid(column=0, row=5, padx=10, pady=25)
# Componente Combobox
n = tk.StringVar()
escolha = ttk.Combobox(janela, width=27, textvariable=n)
# Adição de itens no Combobox
escolha['values'] = (
' Janeiro', ' Fevereiro', ' Março', ' Abril', ' Maio', ' Junho', ' Julho', ' Agosto', ' Setembro', ' Outubro',
' Novembro', ' Dezembro')
escolha.grid(column=1, row=5)
escolha.current()
janela.mainloop()
```
